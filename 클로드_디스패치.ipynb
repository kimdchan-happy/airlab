# 클로드 디스패치(Claude Dispatch) 사용법

클로드 디스패치는 어디서든(스마트폰, CI/CD, 웹훅 등) Claude Code에 작업을 보내고 자율적으로 실행시키는 기능입니다.

> **요구 사항:** Claude Pro 또는 Max 플랜 필요

## 1. 디스패치란?

- **목적:** 작업을 보내고 결과를 받는 "fire and forget" 방식의 자율 실행
- **동작 방식:**
  1. 스마트폰/웹훅/CI 등에서 작업(task)을 디스패치로 전송
  2. 클로드가 작업 유형을 분석해 라우팅:
     - 개발 작업 → Claude Code 세션 (코드 수정, 버그 픽스, 테스트)
     - 지식 작업 → Cowork 세션 (분석, 작성, 리서치)
  3. 작업 완료 시 결과(PR, 코드 수정, 메모 등)를 푸시 알림으로 전달

### 다른 원격 제어 방식과의 비교

| 방식 | 용도 | 동작 |
|------|------|------|
| **Dispatch** | 외부 트리거에서 단발성 작업 | 상태 없음, 태스크 기반 실행 |
| **Channels** | 지속적 양방향 통신 | 열린 대화, 지속 컨텍스트 |
| **Remote Control** | 터미널 직접 제어 | 인터랙티브 세션 관리 |

## 2. 초기 설정 (약 2분)

1. 최신 **Claude Desktop** 앱 다운로드 (macOS / Windows x64)
2. **Cowork 탭** → **Dispatch** 클릭
3. QR 코드를 **Claude 모바일 앱**으로 스캔
4. Dispatch 활성화 → 폰과 데스크탑 연결 완료

> **주의:** 컴퓨터가 켜져 있고 Claude Desktop이 실행 중이어야 합니다. 클라우드 컴퓨팅이 아닌 원격 제어 방식입니다.

## 3. Claude Agent SDK로 프로그래밍 방식 디스패치

Agent SDK(구 Claude Code SDK)를 사용하면 코드에서 직접 Claude에 작업을 디스패치할 수 있습니다.

### 설치

### 기본 사용법 (Python)

```python
import asyncio
from claude_agent_sdk import query, ClaudeAgentOptions

async def main():
    async for message in query(
        prompt="auth.py의 버그를 찾아서 수정해줘",
        options=ClaudeAgentOptions(allowed_tools=["Read", "Edit", "Bash"]),
    ):
        print(message)

asyncio.run(main())
```
